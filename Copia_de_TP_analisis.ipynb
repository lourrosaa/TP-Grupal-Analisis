```python
import pandas as pd

url = "https://raw.githubusercontent.com/lourrosaa/TP-Grupal-Analisis/refs/heads/main/sephora_website_dataset.csv"
df = pd.read_csv(url)
print("--- Primeras 5 filas ---\n")
display(df.head()) #para las primeras 5 filas por defecto
```

```python
print("--- Últimas 5 filas ---\n")
display(df.tail()) #para las últimas 5 filas
```

```python
print("--- Dimensiones (Filas, Columnas) ---\n")
print(df.shape)#cuenta filas y columnas
```

```python
print("--- Tipos de datos por columna ---\n")
display(df.dtypes) #tipo de dato por columna
```

```python
print("--- Cantidad de valores nulos por columna ---\n")
display(df.isnull().sum()) #valores nulos por columna
```

Podemos ver que en este Dataset que importamos de Sephora ningun dato se encuentra vacio, ademas de que la mayoria de datos son de texto o de numeros enteros, con solo pocas excepciones.

```python
#Creamos una nueva columna "es_edicion_limitada" en el data frame para identificar a los pro
if 'limited_edition' in df.columns:
    df['es_edicion_limitada'] = df['limited_edition'].astype(int)
else:
    df['es_edicion_limitada'] = df['name'].str.contains('Limited Edition|Limited', case=False, na=False).astype(int)
```

```python
df[["limited_edition", "rating", "love"]].describe()
print(df[["limited_edition", "rating", "love"]].describe())
df["limited_edition"].value_counts()
```

De los 9168 productos, solo 842 son de edicion limitada

```python
df.groupby("limited_edition")["love"].median()
```

De los productos en total, se ve que los normales gustan mas que los productos de edicion limitada. Sin embargo tal vez eso no sea porque sean menos queridos por los compradores, se podria deber tambien a que al ser ediciones limitadas y durar menos tiempo en el mercado tienen menos tiempo para acumular "loves", no todos alcanzan a comprarlos o simplemente las personas prefieren los productos clasicos.

```python
df_rating = df[df["rating"] > 0]
print(df_rating)
df_rating.groupby("limited_edition")["rating"].median()
print(df_rating.groupby("limited_edition")["rating"].median())
```

Aca vemos que no solo las ediciones limitadas tienen menos love que los productos normales, sino que tiene el mismo promedio de rating. Asi que por ahora no vemos que las ediciones limitadas gusten mas que las regulares.
